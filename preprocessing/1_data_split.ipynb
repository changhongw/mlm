```python
import DALI as dali_code
from langdetect import detect, DetectorFactory
from tqdm import tqdm
import pandas as pd
import os

DetectorFactory.seed = 20250101  # fix seed for reproducibility
```

```python
dali_data_path = '../../data/DALI/DALI_v2.0/annot_tismir'
dali_data = dali_code.get_the_DALI_dataset(dali_data_path, skip=[], keep=[])
song_ids = list(dali_data.keys())

English_songs = []
for id in tqdm(song_ids):
    entry = dali_data[id]
    entry_meta = entry.info['metadata']

    paras = entry.annotations['annot']['paragraphs']

    song_lyrics = []
    for para in paras:
        song_lyrics.append(para['text'])
    song_lyrics = ' '.join(song_lyrics)

    languages_detected = detect(song_lyrics)
    
    if entry_meta['language'] == 'english' and languages_detected == 'en':
        English_songs.append(id)

print(f"Total English songs: {len(English_songs)}")
```
