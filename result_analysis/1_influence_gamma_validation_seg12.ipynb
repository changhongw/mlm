```python
import matplotlib.pyplot as plt
plt.rcParams['figure.constrained_layout.use'] = True
import pandas as pd
```

```python
data_dir = "../outputs/valloss_monitor/pred_HitK_seg12_top5_gamma"

val_seg12_gamma0001 = pd.read_csv(f"{data_dir}0.001.csv")
val_seg12_gamma001 = pd.read_csv(f"{data_dir}0.01.csv")
val_seg12_gamma01 = pd.read_csv(f"{data_dir}0.1.csv")
val_seg12_gamma1 = pd.read_csv(f"{data_dir}1.0.csv")
val_seg12_gamma10 = pd.read_csv(f"{data_dir}10.0.csv")
```

```python
df = pd.DataFrame(columns=["gamma", "Hit@1", "Hit@3", "Hit@5"])
df.loc[len(df)] = ["0.001"] + list(val_seg12_gamma0001.iloc[0].values[:3])
df.loc[len(df)] = ["0.01"] + list(val_seg12_gamma001.iloc[0].values[:3])
df.loc[len(df)] = ["0.1"] + list(val_seg12_gamma01.iloc[0].values[:3])
df.loc[len(df)] = ["1"] + list(val_seg12_gamma1.iloc[0].values[:3])
df.loc[len(df)] = ["10"] + list(val_seg12_gamma10.iloc[0].values[:3])
df
```

```python
save_dir = "../outputs"

# plot the results, gamma on x-axis, hit@k on y-axis
plt.figure(figsize=(3.5, 2.1))
plt.plot(df["gamma"], df["Hit@1"], marker='o', label='Hit@1%')
plt.plot(df["gamma"], df["Hit@3"], marker='o', label='Hit@3%')
plt.plot(df["gamma"], df["Hit@5"], marker='o', label='Hit@5%')
plt.xlim(-0.2, 4.2)
plt.xlabel('$\\gamma$')
plt.legend(loc='upper right')
plt.grid()
plt.savefig(f"{save_dir}/influence_gamma_validation_seg12.pdf")
```
